```python
#input
nama=input('Nama Pembeli: ')
no_hp=input('No. Handphone: ')
jurusan=input('Jurusan [SBY/BL/LMP]: ').upper()
jumlah=int(input('Masukkan Jumlah Beli: '))

#proses
#menentukan tujuan
if jurusan== 'SBY':
    tujuan= 'Surabaya'
    harga= 300000
elif jurusan == 'BL':
    tujuan= 'Bali'
    harga= 350000
else:
    tujuan= 'Lampung'
    harga= 500000

#menentukan diskon/potongan harga
if jumlah >=3:
    potongan= (jumlah * harga)*0.1
else:
    potongan= 0

#menentukan total harga
total= (jumlah * harga) - potongan

#output
print('---------------------------------------------------------')
print('PENJUALAN TIKET BUS'.center(50))
print('XYZ'.center(50))
print('---------------------------------------------------------')
print('Nama Pembeli     : ', nama)
print('No. Handphone    : ', no_hp)
print('Kode Jurusan     : ', jurusan)
print('Nama Kota Tujuan : ', tujuan)
print('Harga            : ', harga)
print('Jumlah Beli      : ', jumlah)
print('--------------------------------------------------------')
print('Potongan harga yang didapat: ', potongan)
print('Total Bayar                : ', total)
uang_bayar=int(input('Uang Bayar               : '))
uang_kembali= uang_bayar - total
print('Kembalian                  : ', uang_kembali)
```
