```python
#setiap transaksi perlu menginput data calon mahasiswa seperti NIS,nama,jurusan
#bagian tabel dijabarkan sbb:
#Kode Jurusan: SI dan SIA
#Nama Prodi: Sistem Informasi dan Sistem Informasi Akuntansi
#Harga: 2400000 dan 2000000
print('==========Data Calon Mahasiswa==========')
nis=input('NIS: ')
nama_mhs=input('Nama Mahasiswa: ')
jurusan=input('Jurusan: ')
print('========================================')

if jurusan== 'Sistem Informasi':
    kode= 'SI'
    harga= 2400000
elif jurusan== 'Sistem Informasi Akuntansi':
    kode= 'SIA'
    harga= 2000000
else:
    kode= 'Jurusan Tidak Tersedia'
    harga= 0

print('NIS            : ', nis)
print('Nama Mahasiswa : ', nama_mhs)
print('Jurusan        : ', jurusan)
print('Kode Jurusan   : ', kode)
print('Biaya Kuliah   : ', harga)
```
