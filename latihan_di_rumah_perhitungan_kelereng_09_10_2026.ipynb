```python
#Aldi mempunyai kelereng 15 butir lebih banyak daripada Budi
#Aldi (a) = Budi (b) + 15
#Anto memiliki kelereng 2x dari kelereng Budi dan Aldi
#Anto (c) = 2 x a + b
#Agung memiliki kelereng 5 butir lebih sedikit daripada Aldi, Budi dan Anto
#Agung (d)= (a+b+c) - 5
#Berapa jumlah kelereng Budi, Anto dan Agung apabila jumlah kelereng Aldi diketahui
print('===============================================================')
#a= b+15
#b= a-15
#jadi a= a-15+15 = a
#c= 2*(a+b)
#d=(a+b+c)-5
a=int(input('Masukkan jumlah kelereng Aldi: '))
b=a-15
c=2*(a+b)
d=(a+b+c) - 5
print('===============================================================')
print('Jumlah kelereng Aldi: ', a)
print('Jumlah kelereng Budi: ', b)
print('Jumlah kelereng Anto: ', c)
print('Jumlah kelereng Agung: ', d)
```
